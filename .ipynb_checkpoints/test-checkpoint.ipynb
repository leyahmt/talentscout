```python
import tkinter as tk
from tkinter import messagebox, ttk
import sqlite3
import datetime
import pandas as pd
from email_validator import validate_email, EmailNotValidError

DB_NAME='LMS_v3.db'
CSV_NAME='LMS_applications_output.csv'
def init_db():
    """Initializes the database safely without erasing existing entries on app restart."""
    with sqlite3.connect(DB_NAME) as conn:
        cursor=conn.cursor()
        cursor.execute("""
            CREATE TABLE IF NOT EXISTS applications (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                app_date TEXT, name TEXT, age INTEGER, gender TEXT, city TEXT, 
                phone TEXT, email TEXT, employment_type TEXT, cibil INTEGER, 
                income REAL, loan_type TEXT, amt REAL, tenure_months INTEGER, 
                interest_rate REAL, emi REAL, total_payable REAL, dti_ratio REAL, 
                status TEXT, rejection_reason TEXT)
        """)
        conn.commit()

def validate_phone(phone):
    return phone.isdigit() and phone.startswith(("6","7","8","9")) and len(phone)==10
def check_email_valid(email):
    try:
        validate_email(email, check_deliverability=False)
        return True
    except EmailNotValidError:
        return False
def calc_emi(principal, annual_rate, tenure_months):
    r=(annual_rate/12)/100
    if r==0:
        return round(principal/tenure_months, 2)
    emi=principal*(r*(1+r)**tenure_months)/((1+r)**tenure_months - 1)
    return round(emi, 2)

def save_to_db(data):
    """Saves a unified transaction record to the database."""
    with sqlite3.connect(DB_NAME) as conn:
        cursor=conn.cursor()
        cursor.execute("""
            INSERT INTO applications VALUES (NULL,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?,?)
        """, data)
        conn.commit()

def process_loan_app():
    lbl_result.config(text="")
    name, phone, email, city=entry_name.get().strip(), entry_phone.get().strip(), entry_email.get().strip(), entry_city.get().strip()
    gender, emp_type, loan_type=gender_var.get(), emp_var.get(), loan_var.get()
    if not all([name, phone, email, city]):
        messagebox.showerror("INPUT ERROR!", "All fields are required!")
        return
    if not validate_phone(phone):
        messagebox.showerror("VALIDATION ERROR!", "Invalid 10-digit phone number!")
        return
    if not check_email_valid(email):
        messagebox.showerror("VALIDATION ERROR!", "Invalid email format!")
        return

    try:
        age=int(entry_age.get())
        cibil=int(entry_cibil.get())
        income=float(entry_income.get())
        amt=float(entry_loan.get())
        tenure=int(entry_tenure.get())
    except ValueError:
        messagebox.showerror("INPUT ERROR!", "Ensure Age, CIBIL, Income, Amount, and Tenure are valid numbers!")
        return

    app_date=datetime.date.today().strftime("%Y-%m-%d")
    status, reason="Approved", ""
    base_rate, final_rate=10.0, 10.0
    emi, dti, total_payable=0.0, 0.0, 0.0

    if 750<=cibil<=900:
        final_rate=base_rate
    elif 700<=cibil<=749:
        final_rate=base_rate+1.5
    elif 650<=cibil<=699:
        final_rate=base_rate+3.0
    else:
        status, reason="Rejected", "Low CIBIL Score"

    if status=="Approved":
        emi=calc_emi(amt, final_rate, tenure)
        dti=round((emi/income)*100, 2)
        total_payable=round(emi*tenure, 2)
        if emi>(income*0.5):
            status, reason="Rejected", "EMI>50% of income"

    record=(app_date, name, age, gender, city, phone, email, emp_type, cibil, 
              income, loan_type, amt, tenure, final_rate if status=="Approved" else 0, 
              emi, total_payable, dti, status, reason if reason else None)
    save_to_db(record)
    if status=="Rejected":
        lbl_result.config(text=f"STATUS: Rejected\nReason: {reason}", fg="#D32F2F")
        messagebox.showerror("REJECTED!", f"Application Rejected:\n{reason}")
    else:
        lbl_result.config(text=f"STATUS: Approved\nRate: {final_rate}%\nEMI: ₹{emi:,.2f}\nTotal: ₹{total_payable:,.2f}", fg="#1B7F2E")
        messagebox.showinfo("APPROVED!", f"Loan approved successfully!\nEMI: ₹{emi:,.2f}")

def export_csv():
    with sqlite3.connect(DB_NAME) as conn:
        df=pd.read_sql_query("SELECT*FROM applications ORDER BY id DESC", conn)
    if df.empty:
        messagebox.showwarning("No Data", "There are no applications inside the database to export yet.")
        return
    df.to_csv(CSV_NAME, index=False)
    print(f"Exported! {df.shape}")
    messagebox.showinfo("Exported!", f"Data successfully saved to '{CSV_NAME}'\nRows/Columns Found: {df.shape}")

init_db()
root=tk.Tk()
root.title("Loan Management System")
root.geometry("480x660")
root.resizable(False, False)
tk.Label(root, text="Loan Application Portal", font=("Helvetica", 18, "bold")).pack(pady=15)
frame=tk.Frame(root)
frame.pack(pady=5)

def add_row(label_text, widget, r):
    tk.Label(frame, text=label_text, font=("Helvetica", 11)).grid(row=r, column=0, sticky="w", pady=5, padx=10)
    widget.grid(row=r, column=1, pady=5, padx=10, sticky="ew")
    
entry_name, entry_phone, entry_email, entry_city=tk.Entry(frame, width=28), tk.Entry(frame, width=28), tk.Entry(frame, width=28), tk.Entry(frame, width=28)
entry_age, entry_cibil, entry_income, entry_loan, entry_tenure=tk.Entry(frame, width=28), tk.Entry(frame, width=28), tk.Entry(frame, width=28), tk.Entry(frame, width=28), tk.Entry(frame, width=28)

gender_var, emp_var, loan_var=tk.StringVar(value="Male"), tk.StringVar(value="Salaried"), tk.StringVar(value="Personal Loan")

add_row("Full Name:", entry_name, 0)
add_row("Age:", entry_age, 1)
add_row("Phone (+91):", entry_phone, 2)
add_row("Email:", entry_email, 3)
add_row("City:", entry_city, 4)
add_row("Gender:", ttk.Combobox(frame, textvariable=gender_var, values=["Male","Female","Other"], state="readonly", width=26), 5)
add_row("Employment:", ttk.Combobox(frame, textvariable=emp_var, values=["Salaried","Self-Employed","Business Owner","Government"], state="readonly", width=26), 6)
add_row("Loan Type:", ttk.Combobox(frame, textvariable=loan_var, values=["Personal Loan","Home Loan","Car Loan","Education Loan","Business Loan"], state="readonly", width=26), 7)
add_row("CIBIL Score:", entry_cibil, 8)
add_row("Monthly Income (₹):", entry_income, 9)
add_row("Loan Amount (₹):", entry_loan, 10)
add_row("Tenure (months):", entry_tenure, 11)

tk.Button(root, text="Submit Application", command=process_loan_app, bg="#C80972", fg="white", font=("Helvetica", 11, "bold"), padx=15, pady=5).pack(pady=10)
lbl_result=tk.Label(root, text="", font=("Helvetica", 12, "bold"), justify="left")
lbl_result.pack(pady=5)
tk.Button(root, text="Export to CSV (Power BI)", command=export_csv, bg="#1A237E", fg="white", font=("Helvetica", 10), padx=10, pady=4).pack(pady=10)
root.mainloop()
```
